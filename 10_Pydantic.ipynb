```python
from pydantic import BaseModel
```

```python
class employee_schema(BaseModel):
    name:str
    id:int
    department:str
```

```python
obj = employee_schema(name="Amarnatha",id=1,department="IT")
obj
```

```python
obj2 = employee_schema(**{"name": "Amarnatha", "id": 1, "department": 101})
obj2                          
```

StrictTpyes


```python
from pydantic import StrictInt
class employee_schema(BaseModel):
    name:str
    id:StrictInt
    department:str
obj3 = employee_schema(**{"name": "Amarnatha", "id": "123", "department": "IT"})
obj3
    
```

2_Fields

```python
from pydantic import BaseModel, Field, EmailStr
from typing import Optional, Literal

class employee_schema(BaseModel):

    name: str = Field(..., min_length=3, max_length=50)
    age: int = Field(..., gt=18, lt=65)
    email: EmailStr = Field(..., description="Employee's email address")
    department: Optional[str] = None
    gender: Literal['Male', 'Female', 'Other'] = Field(..., description="Employee's gender")
obj = employee_schema(name="John Doe", age=30, email="john.doe@example.com", department="IT", gender="Male")
obj
```
